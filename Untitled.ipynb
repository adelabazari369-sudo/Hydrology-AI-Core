```python
import pandas as pd
import numpy as np
print("Core tools are ready for action.")
```

```python
# Create a mock CAMELS dataset for 10 days
dates = pd.date_range(start="2024-10-01", periods=10)
precipitation = np.random.uniform(0, 20, 10)  # بارش به میلی‌متر
temperature = np.random.uniform(5, 15, 10)    # دما به سانتی‌گراد
streamflow = (precipitation * 0.5) + np.random.uniform(0, 2, 10)  # دبی شبیه‌سازی شده

# ساختن جدول داده‌ها (دیتافریم)
data = {
    'Precipitation_mm': np.round(precipitation, 2),
    'Temperature_C': np.round(temperature, 2),
    'Streamflow_m3s': np.round(streamflow, 2)
}

df = pd.DataFrame(data, index=dates)

print("Mock CAMELS Dataset for Basin X:")
df
```

```python
import matplotlib.pyplot as plt

# تنظیم اندازه نمودار
plt.figure(figsize=(10, 5))

# رسم خط بارش و دبی
plt.plot(df.index, df['Precipitation_mm'], label='Precipitation (mm)', color='blue', marker='o', linestyle='--')
plt.plot(df.index, df['Streamflow_m3s'], label='Streamflow (m3/s)', color='red', marker='s', linewidth=2)

# زیباسازی نمودار
plt.title('Precipitation vs Streamflow - Hydrology AI', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Values (mm or m3/s)', fontsize=12)
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

# نمایش خروجی
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

# احضار ماشین فشرده‌ساز
scaler = MinMaxScaler()

# فشرده‌سازی تمام داده‌های جدول
scaled_data = scaler.fit_transform(df)

# ساخت جدول جدید با داده‌های فشرده شده
scaled_df = pd.DataFrame(scaled_data, columns=df.columns, index=df.index)

print("Data successfully normalized for Deep Learning:")
scaled_df
```

```python
# تابعی برای ساخت پنجره‌های زمانی
def create_sequences(data, lookback_days):
    X_seq, Y_seq = [], []
    for i in range(len(data) - lookback_days):
        # برداشتن دیتای 3 روز متوالی بارش و دما (ورودی)
        window = data.iloc[i:(i + lookback_days)][['Precipitation_mm', 'Temperature_C']].values
        # برداشتن دبیِ روزِ بعد از آن 3 روز (هدف)
        target = data.iloc[i + lookback_days]['Streamflow_m3s']
        X_seq.append(window)
        Y_seq.append(target)
    return np.array(X_seq), np.array(Y_seq)

# تعیین طول حافظه مدل روی 3 روز
lookback = 3
X_train, Y_train = create_sequences(scaled_df, lookback)

print(f"Shape of X_train (Input Matrix): {X_train.shape}")
print(f"Shape of Y_train (Target Vector): {Y_train.shape}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# 1. استارت ساخت اسکلت شبکه
model = Sequential()

# 2. اضافه کردن لایه LSTM (مغز اصلی با 50 سلول عصبی)
# توجه کن که input_shape دقیقاً همان دو عدد آخر ماتریس توست: 3 روز و 2 ویژگی
model.add(LSTM(units=50, activation='relu', input_shape=(3, 2)))

# 3. لایه خروجی (چون فقط یک متغیر یعنی دبی را پیش‌بینی می‌کنیم، فقط 1 نورون دارد)
model.add(Dense(units=1))

# 4. کامپایل کردن مدل (استخدام یک مدیر برای اصلاح خطاهای مدل)
model.compile(optimizer='adam', loss='mse')

print("LSTM Brain is successfully created!")
model.summary()
```

```python
print("Initiating Model Training (Deep Learning Phase)...")

# آموزش مدل با 50 تکرار (Epoch)
history = model.fit(X_train, Y_train, epochs=50, verbose=0)

print("Training Completed!")

# رسم نمودار خطای مدل در طول آموزش
plt.figure(figsize=(8, 4))
plt.plot(history.history['loss'], color='purple', linewidth=2)
plt.title('Model Learning Curve (Error Reduction)', fontweight='bold')
plt.xlabel('Epochs (Number of times the model saw the data)')
plt.ylabel('Error (MSE)')
plt.grid(True, linestyle=':', alpha=0.7)
plt.show()
```

```python
# 1. گرفتن آزمون از مدل (پیش‌بینی آینده)
print("Generating AI Predictions...")
predictions = model.predict(X_train)

# 2. رسم نمودار مقایسه واقعیت با پیش‌بینی هوش مصنوعی
plt.figure(figsize=(10, 5))
plt.plot(Y_train, label='Actual Streamflow (Reality)', color='blue', marker='o', linewidth=2)
plt.plot(predictions, label='AI Predicted Streamflow', color='red', marker='x', linestyle='--', linewidth=2)

plt.title('AI Prediction vs Reality', fontsize=14, fontweight='bold')
plt.xlabel('Time Steps (Days)', fontsize=12)
plt.ylabel('Scaled Streamflow', fontsize=12)
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)
plt.show()
```

```python
# ذخیره کردن کل معماری و وزن‌های شبکه عصبی در یک فایل
model.save("Hydrology_LSTM_Brain_v1.keras")

print("AI Brain secured and saved to disk.")
```
